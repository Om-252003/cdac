```python
import numpy as numpy
import cv2
import matplotlib.pyplot as plt
```

```python
denis = cv2.imread('Denis_Mukwege.jpg', 0)
plt.figure(figsize=(5,5))
plt.imshow(denis, cmap='gray')
```

```python
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
face_rects = face_cascade.detectMultiScale(denis)
face_rects #x,y,w,h
```

```python
deniscopy = denis.copy()
for (x,y,w,h) in face_rects:
  cv2.rectangle(denis, (x,y), (x+w,y+h), (255,255,255), 5)
plt.imshow(deniscopy, cmap='gray')
```

```python
group = cv2.imread('group.jpg')
plt.figure(figsize=(5,5))
plt.imshow(group)

face_rects = face_cascade.detectMultiScale(group)
print(face_rects)

groupcopy = group.copy()
for x,y,w,h in face_rects:
  cv2.rectangle(groupcopy, (x,y), (x+w,y+h), (255,255,255), 2)
plt.imshow(groupcopy)
```

```python
group_contour = cv2.imread('group.jpg')
plt.figure(figsize=(5,5))
plt.imshow(group_contour)

gray_group = cv2.cvtColor(group_contour, cv2.COLOR_BGR2RGB)

edges = cv2.Canny(gray_group, 50, 200)

contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw contours on the original color image
# -1 means draw all contours
cv2.drawContours(group_contour, contours, -1, (0,255,0), 1)

plt.figure(figsize=(8,8))
plt.imshow(group_contour)
```

```python
group = cv2.imread('group.jpg')
plt.figure(figsize=(5,5))
plt.imshow(group)

face_rects = face_cascade.detectMultiScale(group, scaleFactor=1.2, minNeighbors=5)
print(face_rects)

groupcopy = group.copy()
for x,y,w,h in face_rects:
  cv2.rectangle(groupcopy, (x,y), (x+w,y+h), (255,255,255), 2)
plt.imshow(groupcopy)
```
