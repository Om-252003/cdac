```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# binary cross entropy loss function
# -(y log(yhat) + (1 - y) log(1 - yhat))

def binary_cross_entropy(y, yhat):
  if y == 1:
    return -np.log(yhat)
  else:
    return -np.log(1 - yhat)
```

```python
yhat = np.arange(0.1, 1.1, 0.1)
yhat
```

```python
print(yhat, binary_cross_entropy(1, yhat))
# means we're expecting the 1 as expected output and yhat is predicted

plt.xlabel('yhat')
plt.ylabel('loss, because actual value expected is 1')
plt.plot(yhat, binary_cross_entropy(1, yhat), c="m",marker='o', mfc='y')
plt.xticks(yhat)
```
