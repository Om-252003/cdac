```python
import numpy as np
```

```python
# user defined or custom word embedding
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#define data
train_data= ['Well done','Good Work','Great effort','Nice Work','Excellent','Weak','Poor effort','Not good','Very Very Bad']
train_label= [1,1,1,1,1,0,0,0,0]
```

```python
# Embedding layer must be first layer
```

```python
max_len = 0
my_data=set()
for w in train_data:
  s=w.split()
  if len(s) > max_len:
    max_len = len(s)
  for w in s:
    my_data.add(w.lower())

print(max_len)
print(my_data)
```

```python
num_tokens = len(my_data) + 1  # token starts from 0
print(num_tokens)
```

```python
input_words=sorted(list(my_data))
input_words_dict={w:i+1 for i,w in enumerate(my_data)}
```

```python
enocded_input_data=np.zeros((len(train_data),max_len))
for i in range(len(train_data)):
  for pos , w in enumerate(train_data[i].split()):
    enocded_input_data[i,pos]=input_words_dict[w.lower()]
```

```python
enocded_input_data
```

```python
from tensorflow.keras.layers import Embedding, Input, LSTM, Dense, Flatten

from tensorflow.keras.models import Sequential
```

```python
model = Sequential()
dim = 512
model.add(Input(shape=(4,)))
model.add(Embedding(num_tokens, dim))

model.add(Flatten())
model.add(Dense(1, activation='sigmoid'))
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])

model.summary()
```

```python
model.fit(enocded_input_data, train_label, epochs=25)
```
