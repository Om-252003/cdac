```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from tensorflow.keras.preprocessing import image
```

```python
img = image.load_img('/content/sunset.jpg' ,color_mode='grayscale')
img_arr = np.asarray(img)
img_arr.shape
```

```python
new_img = np.zeros((img_arr.shape[0]-2,
                    img_arr.shape[1]-2))
new_img.shape
```

```python
plt.imshow(new_img , cmap='gray')
```

```python
gx = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])

gy = gx.T  # OR gy = np.transpose(gx)
```

```python
for i in range(new_img.shape[0]):
  for j in range(new_img.shape[1]):
    x = np.tensordot(img_arr[i:i+3, j:j+3], gx)
    y = np.tensordot(img_arr[i:i+3, j:j+3], gy)
    new_img[i, j] = np.sqrt(x**2 + y**2)**0.5
plt.imshow(new_img , cmap='gray')
```
